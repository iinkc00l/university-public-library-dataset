```python
# ============================================================
# DATA PREPROCESSING FOR RECOMMENDER SYSTEM
#
# OUTPUTS:
#   1. df_clean : cleaned transaction-level dataset
#   2. R0       : initial binary user-item interaction matrix
#
# MAIN AUDIT INFORMATION:
#   - raw rows
#   - malformed CSV rows
#   - missing user IDs
#   - missing item IDs
#   - missing Borrowed values
#   - invalid Borrowed values
#   - negative Borrowed values
#   - duplicate transaction rows
#   - duplicate user-item interactions
#   - total dropped records
#   - final retained records
#   - unique users/items
#   - R0 dimensions
#   - positive interactions
#   - sparsity/density
#
# CORE VARIABLES:
#   User ID : Membership Number
#   Item ID : Book ID Number
#   Feedback: Borrowed
# ============================================================


# ============================================================
# CELL 1 — INSTALL AND IMPORT LIBRARIES
# ============================================================

!pip -q install pandas numpy scipy chardet openpyxl

import os
import gc
import chardet
import numpy as np
import pandas as pd

from scipy import sparse

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)

SEED = 42
np.random.seed(SEED)

print("✅ Libraries loaded.")


# ============================================================
# CELL 2 — DATASET PATH
# ============================================================

path = "/content/drive/MyDrive/Data/dataset_univLib.csv"

assert os.path.exists(path), f"❌ File not found: {path}"

print("✅ Dataset found:")
print(path)


# ============================================================
# CELL 3 — HELPER FUNCTION FOR COLUMN IDENTIFICATION
# ============================================================

def pick_col(df, candidates):
    """
    Find a column even when alternative capitalization/naming is used.
    """

    # exact match
    for c in candidates:
        if c in df.columns:
            return c

    # case-insensitive match
    lower_map = {str(c).lower(): c for c in df.columns}

    for c in candidates:
        if str(c).lower() in lower_map:
            return lower_map[str(c).lower()]

    return None


# ============================================================
# CELL 4 — DETECT FILE ENCODING
# ============================================================

with open(path, "rb") as f:
    raw_sample = f.read(200_000)

encoding_result = chardet.detect(raw_sample)

enc_guess = encoding_result.get("encoding")
enc_confidence = encoding_result.get("confidence")

print("\n" + "=" * 70)
print("ENCODING DETECTION")
print("=" * 70)

print(f"Detected encoding  : {enc_guess}")
print(f"Confidence         : {enc_confidence}")


# ============================================================
# CELL 5 — ROBUST CSV LOADING WITH BAD-LINE AUDIT
# ============================================================

candidate_encodings = []

for enc in [
    enc_guess,
    "utf-8-sig",
    "utf-8",
    "utf8",
    "latin1",
    "ISO-8859-1",
    "Windows-1252"
]:
    if enc and enc not in candidate_encodings:
        candidate_encodings.append(enc)


df_raw = None
successful_encoding = None
bad_lines_final = []


for enc in candidate_encodings:

    bad_lines_temp = []

    def capture_bad_line(fields):
        """
        Save malformed CSV rows, then skip them.
        """
        bad_lines_temp.append(fields)
        return None

    try:

        temp_df = pd.read_csv(
            path,
            encoding=enc,
            sep=None,
            engine="python",
            on_bad_lines=capture_bad_line
        )

        df_raw = temp_df
        successful_encoding = enc
        bad_lines_final = bad_lines_temp

        print(f"\n✅ Successfully read dataset using encoding: {enc}")
        break

    except Exception as e:

        print(f"⚠️ Failed with encoding {enc}")
        print(f"   Reason: {str(e)[:120]}")


if df_raw is None:
    raise RuntimeError(
        "❌ Dataset could not be read using available encodings."
    )


print("\n" + "=" * 70)
print("RAW DATASET")
print("=" * 70)

print(f"Rows successfully parsed : {len(df_raw):,}")
print(f"Columns                  : {df_raw.shape[1]:,}")
print(f"Malformed rows skipped   : {len(bad_lines_final):,}")
print(f"Encoding used            : {successful_encoding}")

N_RAW_PARSED = len(df_raw)


# ============================================================
# CELL 6 — IDENTIFY CORE RECOMMENDER COLUMNS
# ============================================================

user_col = pick_col(
    df_raw,
    [
        "Membership Number",
        "membership_number",
        "MembershipNumber"
    ]
)

# IMPORTANT:
# Book ID Number is the ITEM identifier
item_col = pick_col(
    df_raw,
    [
        "Book ID Number",
        "book_id_number",
        "BookIDNumber"
    ]
)

borrow_col = pick_col(
    df_raw,
    [
        "Borrowed",
        "borrowed",
        "rating",
        "Rating",
        "label",
        "Label"
    ]
)


print("\n" + "=" * 70)
print("CORE COLUMN IDENTIFICATION")
print("=" * 70)

print(f"User column        : {user_col}")
print(f"Item column        : {item_col}")
print(f"Interaction column : {borrow_col}")


if user_col is None:
    raise ValueError(
        "❌ Membership Number could not be identified."
    )

if item_col is None:
    raise ValueError(
        "❌ Book ID Number could not be identified."
    )

if borrow_col is None:
    raise ValueError(
        "❌ Borrowed column could not be identified."
    )


# ============================================================
# CELL 7 — INITIAL MISSING-VALUE AUDIT
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUE AUDIT — RAW DATA")
print("=" * 70)


raw_missing_user = df_raw[user_col].isna()
raw_missing_item = df_raw[item_col].isna()
raw_missing_borrow = df_raw[borrow_col].isna()


print(f"Missing {user_col:<25}: {raw_missing_user.sum():>10,}")
print(f"Missing {item_col:<25}: {raw_missing_item.sum():>10,}")
print(f"Missing {borrow_col:<25}: {raw_missing_borrow.sum():>10,}")


missing_table = pd.DataFrame({
    "Column": df_raw.columns,
    "Missing_Count": df_raw.isna().sum().values,
    "Missing_Percent": (
        df_raw.isna().sum().values
        / len(df_raw)
        * 100
    )
})

missing_table = (
    missing_table
    .sort_values(
        "Missing_Count",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nMissing-value summary for all variables:")
display(missing_table)


# ============================================================
# CELL 8 — MAKE WORKING COPY
# ============================================================

df_work = df_raw.copy()

print(f"\nInitial working records: {len(df_work):,}")


# ============================================================
# CELL 9 — STANDARDIZE USER AND ITEM IDs
# ============================================================

# Important:
# Do NOT immediately use astype(str),
# because missing values would become literal "nan".

for c in [user_col, item_col]:

    df_work[c] = df_work[c].astype("string")

    df_work[c] = (
        df_work[c]
        .str.strip()
    )

    # convert blank text to missing
    df_work[c] = df_work[c].replace(
        {
            "": pd.NA,
            " ": pd.NA,
            "nan": pd.NA,
            "None": pd.NA,
            "NULL": pd.NA,
            "null": pd.NA
        }
    )


missing_user_mask = df_work[user_col].isna()
missing_item_mask = df_work[item_col].isna()


print("\n" + "=" * 70)
print("IDENTIFIER QUALITY AFTER STANDARDIZATION")
print("=" * 70)

print(
    f"Missing/blank user IDs : "
    f"{missing_user_mask.sum():,}"
)

print(
    f"Missing/blank item IDs : "
    f"{missing_item_mask.sum():,}"
)


# ============================================================
# CELL 10 — VALIDATE BORROWED VARIABLE
# ============================================================

# Preserve original interaction value
df_work["_Borrowed_original"] = df_work[borrow_col]


borrow_numeric = pd.to_numeric(
    df_work[borrow_col],
    errors="coerce"
)


# missing values
missing_borrow_mask = (
    df_work[borrow_col].isna()
)


# non-numeric but originally non-missing
invalid_borrow_mask = (
    df_work[borrow_col].notna()
    &
    borrow_numeric.isna()
)


# negative values
negative_borrow_mask = (
    borrow_numeric.notna()
    &
    (borrow_numeric < 0)
)


print("\n" + "=" * 70)
print("BORROWED VARIABLE QUALITY")
print("=" * 70)

print(
    f"Missing Borrowed values     : "
    f"{missing_borrow_mask.sum():,}"
)

print(
    f"Invalid/non-numeric values  : "
    f"{invalid_borrow_mask.sum():,}"
)

print(
    f"Negative Borrowed values    : "
    f"{negative_borrow_mask.sum():,}"
)


if invalid_borrow_mask.sum() > 0:

    print("\nExamples of invalid Borrowed values:")

    display(
        df_work.loc[
            invalid_borrow_mask,
            [
                user_col,
                item_col,
                borrow_col
            ]
        ].head(20)
    )


# ============================================================
# CELL 11 — DEFINE RECORDS THAT CANNOT PROCEED
# ============================================================

# A row cannot proceed if:
#
# 1. Membership Number missing
# 2. Book ID Number missing
# 3. Borrowed missing
# 4. Borrowed non-numeric
# 5. Borrowed negative

invalid_core_mask = (
    missing_user_mask
    |
    missing_item_mask
    |
    missing_borrow_mask
    |
    invalid_borrow_mask
    |
    negative_borrow_mask
)


rows_invalid_core = int(
    invalid_core_mask.sum()
)


print("\n" + "=" * 70)
print("CORE VALIDATION")
print("=" * 70)

print(
    f"Rows failing ≥1 mandatory rule : "
    f"{rows_invalid_core:,}"
)

print(
    f"Rows eligible to continue      : "
    f"{len(df_work) - rows_invalid_core:,}"
)


# keep dropped rows for auditing
df_dropped_core = (
    df_work
    .loc[invalid_core_mask]
    .copy()
)


# remove invalid rows
df_work = (
    df_work
    .loc[~invalid_core_mask]
    .copy()
)


print(
    f"\n✅ Remaining rows after core validation: "
    f"{len(df_work):,}"
)


# ============================================================
# CELL 12 — CONVERT BORROWED TO BINARY IMPLICIT FEEDBACK
# ============================================================

borrow_numeric_clean = pd.to_numeric(
    df_work[borrow_col],
    errors="raise"
)


nonbinary_positive_count = int(
    (borrow_numeric_clean > 1).sum()
)


# Binary implicit feedback:
#
# Borrowed > 0 -> 1
# Borrowed = 0 -> 0

df_work[borrow_col] = (
    borrow_numeric_clean > 0
).astype(np.int8)


print("\n" + "=" * 70)
print("IMPLICIT FEEDBACK TRANSFORMATION")
print("=" * 70)

print(
    f"Values >1 converted to 1     : "
    f"{nonbinary_positive_count:,}"
)

print(
    f"Binary positive rows (1)     : "
    f"{(df_work[borrow_col] == 1).sum():,}"
)

print(
    f"Binary zero rows (0)         : "
    f"{(df_work[borrow_col] == 0).sum():,}"
)


# ============================================================
# CELL 13 — EXACT DUPLICATE TRANSACTION AUDIT
# ============================================================

duplicate_check_cols = [
    c
    for c in df_raw.columns
    if c in df_work.columns
]


exact_duplicate_mask = (
    df_work
    .duplicated(
        subset=duplicate_check_cols,
        keep="first"
    )
)


n_exact_duplicates = int(
    exact_duplicate_mask.sum()
)


print("\n" + "=" * 70)
print("EXACT DUPLICATE TRANSACTION AUDIT")
print("=" * 70)

print(
    f"Exact duplicate rows identified : "
    f"{n_exact_duplicates:,}"
)


df_dropped_duplicates = (
    df_work
    .loc[exact_duplicate_mask]
    .copy()
)


# remove only exact duplicated transaction rows
df_work = (
    df_work
    .loc[~exact_duplicate_mask]
    .copy()
)


print(
    f"Rows after exact duplicate removal : "
    f"{len(df_work):,}"
)


# ============================================================
# CELL 14 — FINAL CLEANED TRANSACTION DATASET
# ============================================================

df_work = df_work.drop(
    columns=[
        "_Borrowed_original"
    ],
    errors="ignore"
)


df_clean = (
    df_work
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("OBJECT 1 — CLEANED TRANSACTION DATASET")
print("=" * 70)

print(
    f"Shape                    : "
    f"{df_clean.shape}"
)

print(
    f"Rows                     : "
    f"{len(df_clean):,}"
)

print(
    f"Columns                  : "
    f"{df_clean.shape[1]:,}"
)

print(
    f"Unique users             : "
    f"{df_clean[user_col].nunique():,}"
)

print(
    f"Unique Book ID Numbers   : "
    f"{df_clean[item_col].nunique():,}"
)

print(
    f"Positive transaction rows: "
    f"{(df_clean[borrow_col] == 1).sum():,}"
)

print(
    f"Zero transaction rows    : "
    f"{(df_clean[borrow_col] == 0).sum():,}"
)


# ============================================================
# CELL 15 — REPEATED USER–BOOK INTERACTION AUDIT
# ============================================================

# A user may borrow the same Book ID Number more than once.
#
# These are legitimate historical transaction events.
#
# They remain in df_clean.
#
# They will be collapsed into a single binary relationship
# when R0 is constructed.

pair_counts = (
    df_clean
    .groupby(
        [
            user_col,
            item_col
        ],
        observed=True
    )
    .size()
    .reset_index(
        name="transaction_count"
    )
)


repeated_pairs = (
    pair_counts[
        pair_counts["transaction_count"] > 1
    ]
    .copy()
)


extra_repeated_transactions = int(
    (
        repeated_pairs["transaction_count"] - 1
    ).sum()
)


print("\n" + "=" * 70)
print("REPEATED USER–BOOK INTERACTION AUDIT")
print("=" * 70)

print(
    f"Unique user-book pairs       : "
    f"{len(pair_counts):,}"
)

print(
    f"Pairs appearing >1 time      : "
    f"{len(repeated_pairs):,}"
)

print(
    f"Extra repeated transactions  : "
    f"{extra_repeated_transactions:,}"
)


if len(repeated_pairs) > 0:

    print("\nExamples of repeated user-book pairs:")

    display(
        repeated_pairs
        .sort_values(
            "transaction_count",
            ascending=False
        )
        .head(20)
    )


# ============================================================
# CELL 16 — BUILD UNIQUE BINARY USER–BOOK INTERACTIONS
# ============================================================

# Multiple transactions for the same user-book pair
# are collapsed using maximum Borrowed value.
#
# If at least one transaction is positive:
#
# R0(u,i) = 1

ui_binary = (
    df_clean
    .groupby(
        [
            user_col,
            item_col
        ],
        observed=True,
        as_index=False
    )[borrow_col]
    .max()
)


print("\n" + "=" * 70)
print("USER–BOOK PAIR COLLAPSING")
print("=" * 70)

print(
    f"Transaction rows in df_clean : "
    f"{len(df_clean):,}"
)

print(
    f"Unique user-book pairs       : "
    f"{len(ui_binary):,}"
)

print(
    f"Positive user-book pairs     : "
    f"{(ui_binary[borrow_col] == 1).sum():,}"
)


# ============================================================
# CELL 17 — CREATE USER AND BOOK INDEX MAPPINGS
# ============================================================

users = pd.Index(
    sorted(
        df_clean[user_col].unique()
    ),
    name=user_col
)


items = pd.Index(
    sorted(
        df_clean[item_col].unique()
    ),
    name=item_col
)


user_to_idx = {
    user: idx
    for idx, user
    in enumerate(users)
}


item_to_idx = {
    item: idx
    for idx, item
    in enumerate(items)
}


user_index_mapping = pd.DataFrame({
    "user_idx": range(len(users)),
    user_col: users
})


item_index_mapping = pd.DataFrame({
    "item_idx": range(len(items)),
    item_col: items
})


print("\n" + "=" * 70)
print("INDEX MAPPINGS")
print("=" * 70)

print(
    f"Unique users          : "
    f"{len(users):,}"
)

print(
    f"Unique Book IDs       : "
    f"{len(items):,}"
)


# ============================================================
# CELL 18 — BUILD R0 AS SPARSE MATRIX
# ============================================================

positive_pairs = (
    ui_binary[
        ui_binary[borrow_col] == 1
    ]
    .copy()
)


row_indices = (
    positive_pairs[user_col]
    .map(user_to_idx)
    .to_numpy()
)


col_indices = (
    positive_pairs[item_col]
    .map(item_to_idx)
    .to_numpy()
)


values = np.ones(
    len(positive_pairs),
    dtype=np.float32
)


R0_sparse = sparse.csr_matrix(
    (
        values,
        (
            row_indices,
            col_indices
        )
    ),
    shape=(
        len(users),
        len(items)
    ),
    dtype=np.float32
)


R0_sparse.sum_duplicates()

R0_sparse.data[:] = 1.0


print(
    "\n✅ Sparse R0 successfully created."
)


# ============================================================
# CELL 19 — R0 MATRIX STATISTICS
# ============================================================

n_users = R0_sparse.shape[0]
n_items = R0_sparse.shape[1]

total_cells = (
    n_users
    *
    n_items
)

filled_cells = (
    R0_sparse.nnz
)

density = (
    filled_cells / total_cells
    if total_cells > 0
    else 0
)

sparsity_value = (
    1 - density
)


print("\n" + "=" * 70)
print("OBJECT 2 — INITIAL USER–ITEM MATRIX R0")
print("=" * 70)

print(
    f"Matrix dimensions        : "
    f"{n_users:,} × {n_items:,}"
)

print(
    f"Total possible cells     : "
    f"{total_cells:,}"
)

print(
    f"Observed positive cells  : "
    f"{filled_cells:,}"
)

print(
    f"Density                  : "
    f"{density:.8f}"
)

print(
    f"Sparsity                 : "
    f"{sparsity_value:.8f}"
)

print(
    f"Sparsity (%)             : "
    f"{sparsity_value * 100:.6f}%"
)


# ============================================================
# CELL 20 — INTERACTION DISTRIBUTION
# ============================================================

user_interaction_counts = np.asarray(
    R0_sparse.sum(axis=1)
).ravel()


item_interaction_counts = np.asarray(
    R0_sparse.sum(axis=0)
).ravel()


# ----------------------------
# USERS
# ----------------------------

zero_interaction_users = int(
    (
        user_interaction_counts == 0
    ).sum()
)

one_interaction_users = int(
    (
        user_interaction_counts == 1
    ).sum()
)

less_than_2_users = int(
    (
        user_interaction_counts < 2
    ).sum()
)


# ----------------------------
# ITEMS
# ----------------------------

zero_interaction_items = int(
    (
        item_interaction_counts == 0
    ).sum()
)

one_interaction_items = int(
    (
        item_interaction_counts == 1
    ).sum()
)

less_than_2_items = int(
    (
        item_interaction_counts < 2
    ).sum()
)


print("\n" + "=" * 70)
print("INTERACTION AVAILABILITY")
print("=" * 70)


print("\nUSERS")

print(
    f"Users with 0 positive interactions : "
    f"{zero_interaction_users:,}"
)

print(
    f"Users with exactly 1 interaction   : "
    f"{one_interaction_users:,}"
)

print(
    f"Users with <2 interactions         : "
    f"{less_than_2_users:,}"
)


print("\nITEMS")

print(
    f"Items with 0 positive interactions : "
    f"{zero_interaction_items:,}"
)

print(
    f"Items with exactly 1 interaction   : "
    f"{one_interaction_items:,}"
)

print(
    f"Items with <2 interactions         : "
    f"{less_than_2_items:,}"
)


# ============================================================
# CELL 21 — COMPLETE PREPROCESSING AUDIT TABLE
# ============================================================

n_final = len(df_clean)

n_total_removed = (
    N_RAW_PARSED
    -
    n_final
)


audit_table = pd.DataFrame({

    "Stage": [

        "Successfully parsed CSV rows",

        "Malformed CSV rows skipped",

        "Missing/blank Membership Number",

        "Missing/blank Book ID Number",

        "Missing Borrowed",

        "Invalid/non-numeric Borrowed",

        "Negative Borrowed",

        "Rows failing one or more core rules",

        "Exact duplicate rows removed",

        "Total rows removed from parsed dataset",

        "Final cleaned transaction rows",

        "Unique users",

        "Unique Book ID Numbers",

        "Unique user-book pairs",

        "Repeated user-book pairs",

        "Positive user-book pairs / R0 filled cells",

        "R0 total possible cells"
    ],

    "Count": [

        N_RAW_PARSED,

        len(bad_lines_final),

        int(
            missing_user_mask.sum()
        ),

        int(
            missing_item_mask.sum()
        ),

        int(
            missing_borrow_mask.sum()
        ),

        int(
            invalid_borrow_mask.sum()
        ),

        int(
            negative_borrow_mask.sum()
        ),

        rows_invalid_core,

        n_exact_duplicates,

        n_total_removed,

        n_final,

        n_users,

        n_items,

        len(ui_binary),

        len(repeated_pairs),

        filled_cells,

        total_cells
    ]
})


audit_table[
    "Percent_of_Parsed_Rows"
] = np.where(

    audit_table["Stage"].isin([

        "Successfully parsed CSV rows",

        "Malformed CSV rows skipped",

        "Unique users",

        "Unique Book ID Numbers",

        "Unique user-book pairs",

        "Repeated user-book pairs",

        "Positive user-book pairs / R0 filled cells",

        "R0 total possible cells"
    ]),

    np.nan,

    (
        audit_table["Count"]
        /
        N_RAW_PARSED
        *
        100
    )
)


print("\n" + "=" * 70)
print("FINAL PREPROCESSING AUDIT")
print("=" * 70)

display(audit_table)


# ============================================================
# CELL 22 — CONSISTENCY CHECKS
# ============================================================

print("\n" + "=" * 70)
print("CONSISTENCY CHECKS")
print("=" * 70)


assert (
    df_clean[user_col]
    .isna()
    .sum()
    == 0
), "User IDs still contain missing values."


assert (
    df_clean[item_col]
    .isna()
    .sum()
    == 0
), "Book ID Number still contains missing values."


assert (
    df_clean[borrow_col]
    .isna()
    .sum()
    == 0
), "Borrowed still contains missing values."


assert set(
    df_clean[borrow_col].unique()
).issubset(
    {0, 1}
), "Borrowed is not binary."


assert (
    R0_sparse.shape
    ==
    (
        df_clean[user_col].nunique(),
        df_clean[item_col].nunique()
    )
), "R0 dimensions do not match unique users/items."


assert (
    R0_sparse.nnz
    ==
    int(
        (
            ui_binary[borrow_col]
            == 1
        ).sum()
    )
), "R0 positive count is inconsistent."


print("✅ No missing Membership Number")
print("✅ No missing Book ID Number")
print("✅ No missing Borrowed values")
print("✅ Borrowed is binary")
print("✅ R0 dimensions are correct")
print("✅ R0 positive-cell count is correct")


# ============================================================
# CELL 23 — DENSE MATRIX MEMORY CHECK
# ============================================================

estimated_dense_gb = (
    total_cells
    *
    4
) / (
    1024 ** 3
)


print("\n" + "=" * 70)
print("DENSE MATRIX MEMORY CHECK")
print("=" * 70)

print(
    f"Estimated raw float32 memory : "
    f"{estimated_dense_gb:.2f} GB"
)


# Sparse representation remains recommended.
#
# Dense representation will only be created automatically
# if estimated memory is <= 1 GB.

MAX_DENSE_GB = 1.0


if estimated_dense_gb <= MAX_DENSE_GB:

    print(
        "\n✅ Matrix is small enough "
        "to create a pandas sparse R0."
    )

    R0 = pd.DataFrame.sparse.from_spmatrix(
        R0_sparse,
        index=users,
        columns=items
    )

    print(
        "✅ R0 pandas sparse DataFrame created."
    )

else:

    R0 = R0_sparse

    print(
        "\n⚠️ Dense R0 was NOT created because "
        "the estimated matrix is too large."
    )

    print(
        "✅ R0 therefore refers to "
        "the scipy CSR sparse matrix."
    )


# ============================================================
# CELL 24 — SAVE PREPROCESSING OUTPUTS
# ============================================================

output_dir = (
    "/content/drive/MyDrive/Data/preprocessing_R0"
)

os.makedirs(
    output_dir,
    exist_ok=True
)


# ----------------------------
# Object 1:
# Cleaned transaction dataset
# ----------------------------

df_clean.to_csv(
    os.path.join(
        output_dir,
        "cleaned_transaction_dataset.csv"
    ),
    index=False
)


# ----------------------------
# Object 2:
# R0 sparse matrix
# ----------------------------

sparse.save_npz(
    os.path.join(
        output_dir,
        "R0_sparse.npz"
    ),
    R0_sparse
)


# ----------------------------
# User index mapping
# ----------------------------

user_index_mapping.to_csv(
    os.path.join(
        output_dir,
        "R0_user_index_mapping.csv"
    ),
    index=False
)


# ----------------------------
# Book index mapping
# ----------------------------

item_index_mapping.to_csv(
    os.path.join(
        output_dir,
        "R0_book_index_mapping.csv"
    ),
    index=False
)


# ----------------------------
# Preprocessing audit
# ----------------------------

audit_table.to_csv(
    os.path.join(
        output_dir,
        "preprocessing_audit.csv"
    ),
    index=False
)


# ----------------------------
# Missing-value summary
# ----------------------------

missing_table.to_csv(
    os.path.join(
        output_dir,
        "missing_value_summary.csv"
    ),
    index=False
)


# ----------------------------
# Invalid core records
# ----------------------------

df_dropped_core.to_csv(
    os.path.join(
        output_dir,
        "dropped_invalid_core_records.csv"
    ),
    index=False
)


# ----------------------------
# Exact duplicate rows
# ----------------------------

df_dropped_duplicates.to_csv(
    os.path.join(
        output_dir,
        "dropped_exact_duplicates.csv"
    ),
    index=False
)


# ----------------------------
# Repeated user-book pairs
# ----------------------------

repeated_pairs.to_csv(
    os.path.join(
        output_dir,
        "repeated_user_book_pairs.csv"
    ),
    index=False
)


print("\n" + "=" * 70)
print("FILES SAVED")
print("=" * 70)

print(
    f"""
📁 {output_dir}

1. cleaned_transaction_dataset.csv
2. R0_sparse.npz
3. R0_user_index_mapping.csv
4. R0_book_index_mapping.csv
5. preprocessing_audit.csv
6. missing_value_summary.csv
7. dropped_invalid_core_records.csv
8. dropped_exact_duplicates.csv
9. repeated_user_book_pairs.csv
"""
)


# ============================================================
# CELL 25 — FINAL SUMMARY
# ============================================================

retention_rate = (
    len(df_clean)
    /
    N_RAW_PARSED
    *
    100
    if N_RAW_PARSED > 0
    else 0
)


removal_rate = (
    n_total_removed
    /
    N_RAW_PARSED
    *
    100
    if N_RAW_PARSED > 0
    else 0
)


print("\n" + "=" * 70)
print("FINAL PREPROCESSING SUMMARY")
print("=" * 70)

print(f"""
RAW DATA
------------------------------------------------------------
Successfully parsed rows       : {N_RAW_PARSED:,}
Malformed CSV rows skipped     : {len(bad_lines_final):,}

CORE DATA QUALITY
------------------------------------------------------------
Missing/blank Membership No.   : {int(missing_user_mask.sum()):,}
Missing/blank Book ID Number   : {int(missing_item_mask.sum()):,}
Missing Borrowed values        : {int(missing_borrow_mask.sum()):,}
Invalid Borrowed values        : {int(invalid_borrow_mask.sum()):,}
Negative Borrowed values       : {int(negative_borrow_mask.sum()):,}
Core-invalid rows removed      : {rows_invalid_core:,}
Exact duplicates removed       : {n_exact_duplicates:,}

OBJECT 1 — CLEANED TRANSACTION DATASET
------------------------------------------------------------
Final transaction rows         : {len(df_clean):,}
Rows removed                   : {n_total_removed:,}
Retention rate                 : {retention_rate:.4f}%
Removal rate                   : {removal_rate:.4f}%

USERS AND BOOKS
------------------------------------------------------------
Unique users                   : {n_users:,}
Unique Book ID Numbers         : {n_items:,}
Unique user-book pairs         : {len(ui_binary):,}
Repeated user-book pairs       : {len(repeated_pairs):,}
Extra repeated transactions    : {extra_repeated_transactions:,}

OBJECT 2 — INITIAL MATRIX R0
------------------------------------------------------------
Dimensions                     : {n_users:,} × {n_items:,}
Positive observed cells        : {filled_cells:,}
Total possible cells           : {total_cells:,}
Density                        : {density:.8f}
Sparsity                       : {sparsity_value:.8f}
Sparsity (%)                   : {sparsity_value * 100:.6f}%

✅ PREPROCESSING COMPLETE
""")
```
